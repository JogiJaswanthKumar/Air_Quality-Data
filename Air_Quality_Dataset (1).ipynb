```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset using the correct absolute path
df = pd.read_csv('/content/Air_Quality.csv')

# --- DATA PREPARATION & AGGREGATION ---

# Filter data for just the City
city_center_data = df[df['Location'] == 'City']

# Group data by location to find average pollution levels (only taking the mean of numeric columns like PM2.5_Level)
avg_pollution = df.groupby('Location')[['PM2.5_Level']].mean()

# --- THE 5 REQUIRED CHARTS ---

# Chart 1: Line Chart (Tracking Air_Quality_Category over time at City)
plt.figure(figsize=(8, 4))

plt.plot(city_center_data['Date'],
         city_center_data['Air_Quality_Category'],
         marker='o', color='red')

plt.title('Chart 1: Air_Quality_Category Trend at City')
plt.xlabel('Date')
plt.ylabel('Air Quality Category')
plt.grid(True)
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by Location)

plt.figure(figsize=(8, 4))

avg_pollution['PM2.5_Level'].plot(kind='bar', color='orange')

plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)

plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 levels over Date)

# Drop any rows where Date is missing to prevent Matplotlib category plotting errors
clean_df = df.dropna(subset=['Date'])

plt.figure(figsize=(8, 4))
plt.scatter(clean_df['Date'], clean_df['PM2.5_Level'], color='purple', s=140)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#char 4
import pandas as pd
import matplotlib.pyplot as plt

# Corrected the filename to match the available file in your workspace
df = pd.read_csv('/content/Air_Quality.csv')

plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_Level'], bins=5, color='skyblue', edgecolor='black')
plt.title('char 4 :Distribution of PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
# Chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='bar', color='blue')
plt.title('char 5 :Average PM2.5 Levels by Location')
plt.xlabel('Peak PM2.5_Level')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='yellow')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Max PM2.5')
plt.xticks(rotation=0)
plt.show()
```
